```python
tables = pd.read_sql(
    "SELECT name FROM sqlite_master WHERE type='table' AND name LIKE 'core_%';",
    conn1
)

counts = []

for table in tables["name"]:
    count = pd.read_sql(
        f"SELECT COUNT(*) AS n FROM {table};",
        conn1
    )["n"][0]
    counts.append({"table": table, "rows": count})

pd.DataFrame(counts).sort_values("rows", ascending=False)
```
