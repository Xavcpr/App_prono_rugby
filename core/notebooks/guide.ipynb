Résumé des relations clés :

    La Colonne Vertébrale (Données Réelles) :

        Competition → possède plusieurs Seasons.

        Season → possède plusieurs Rounds (Journées).

        Round → possède plusieurs Matches.

        Match → lie deux Teams (Home/Away).

    La Structure de Saison :

        CompetitionTeam : C'est le lien "pivot" qui définit quelles Teams jouent dans quelle Season d'une Competition.

    Les Pronostics (Utilisateurs) :

        Prediction : Lié à un Player et à un Match (Score home/away).

        CompetitionTeamPrediction : Le classement final prédit par le joueur pour une Season.

        CompetitionBonusPrediction : Les paris sur le vainqueur final et les meilleurs marqueurs d'une Season.

    Le Scoring (Calculs) :

        DailyScore : Total des points d'un utilisateur pour un Round précis.

        SeasonScore : Cumul total des points (Matchs + Classement) d'un utilisateur pour une Season entière.

        CompetitionResult : Les résultats réels (saisis par toi) qui servent de référence pour calculer les points de fin de saison.

## Descriptif de l'appli

### Flux de fonctionnement 

1. L’utilisateur visite une URL
2. config/urls.py redirige vers core/urls.py
3. core/urls.py appelle une fonction dans views.py
4. views.py récupère les données via models.py
5. views.py retourne pronostics.html
6. Le template affiche les données à l’utilisateur

### config/urls.py

- Fichier de routage principal du projet.
- Il centralise toutes les URLs globales.
- Il redirige les routes vers les applications (ex : core).
ex : path('pronostics/', include('core.urls'))

### core/urls.py

- Fichier de routage spécifique à l’application core.
- Il associe une URL à une vue précise.
- ex : path('', views.pronostics, name='pronostics')

### views.py

- Contient la logique applicative.
- Récupère les données (depuis models.py),
- les traite,
- et les envoie vers un template HTML.

👉 Exemple de rôle :
- charger les pronostics,
- les transmettre à pronostics.html.

### models.py

- Définit la structure de la base de données.
- Chaque classe correspond à une table.
- Django gère ensuite les migrations.
- ex : class Pronostic(models.Model):
    equipe = models.CharField(...)
    score = models.IntegerField(...)

### admin.py

- Permet d’enregistrer les modèles dans l’interface d’administration Django.
- Rend la gestion des données simple via /admin.
- ex : admin.site.register(Pronostic)

### templates/

- Contient les fichiers HTML.
- Utilise le moteur de templates Django ({{ }}, {% %}).

### templates/pronos/
- Page HTML dédiée à l’affichage des pronostics.
- Reçoit des données depuis views.py.
- Affiche dynamiquement les informations.

👉 Exemple de rôle :
- afficher une liste de pronostics,
- utiliser des boucles {% for %}.
